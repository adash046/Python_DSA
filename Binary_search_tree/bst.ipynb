#### Insert 

```python
# Update the above BinarySearchTree class with insert method
class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None

class BinarySearchTree:
    # Initialize the root node
    def __init__(self):
        self.root = None

    # Method to insert a node/value
    def insert(self, value):
        new_node = Node(value)
        # Check if the tree is empty, update the root node itself with the new node
        if self.root is None:
            self.root = new_node
            return True
        # If root node is non empty, then check which side to go 
        temp = self.root
        while True:
            if temp.value == new_node.value:
                return False
            if new_node.value > temp.value:
                if temp.right is None:
                    temp.right = new_node
                    return True
                else:
                    temp = temp.right
            else:
                if temp.left is None:
                    temp.left = new_node
                    return True
                else:
                    temp = temp.left
```

```python
# Create a Binary search tree with one parant node and two child node
my_tree = BinarySearchTree()
my_tree.insert(2)
print(my_tree.root.value)
print(my_tree.root.left)
print(my_tree.root.right)
```

### Contains: This method checks the BST and return True if the (query) value is found in the BST.

```python
# Update the above BinarySearchTree class with "contan" method
class Node:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None

class BinarySearchTree:
    # Initialize the root node
    def __init__(self):
        self.root = None

    # Method to insert a node/value
    def insert(self, value):
        new_node = Node(value)
        # Check if the tree is empty, update the root node itself with the new node
        if self.root is None:
            self.root = new_node
            return True
        # If root node is non empty, then check which side to go 
        temp = self.root
        while True:
            if temp.value == new_node.value:
                return False
            if new_node.value > temp.value:
                if temp.right is None:
                    temp.right = new_node
                    return True
                else:
                    temp = temp.right
            else:
                if temp.left is None:
                    temp.left = new_node
                    return True
                else:
                    temp = temp.left

    # Write a function to implement "contain" method
    def contains(self, value):
        # Cehck if the BST is empty, then return False
        # Not required. One can skip this line of code. The while loop will take care of this
        if self.root is None:
            return False
        # Initialize a temp variable to the root node
        temp = self.root
        while temp is not None:
            if value > temp.value:
                temp = temp.right
            elif value < temp.value:
                temp = temp.left
            else:
                return True
        return False
```

```python
# Create a Binary search tree with one parant node and two child node
my_tree = BinarySearchTree()
my_tree.insert(2)
my_tree.insert(1)
my_tree.insert(3)
```
